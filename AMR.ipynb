```python
import shutil
import os

print(f"Cleaning existing contents of '{combined_dataset_path}'...")
# Remove all contents of the combined dataset directory, but keep the directory itself
for item in os.listdir(combined_dataset_path):
    item_path = os.path.join(combined_dataset_path, item)
    if os.path.isfile(item_path):
        os.remove(item_path)
    elif os.path.isdir(item_path):
        shutil.rmtree(item_path)

print(f"Contents of '{combined_dataset_path}' cleared. Now re-copying data.")

# Re-define the copy_yolov8_data function to handle 'valid' as 'val'
def copy_yolov8_data(source_root, destination_root):
    # Map source splits to destination splits
    split_map = {'train': 'train', 'valid': 'val'} # Source 'valid' will become destination 'val'

    for source_split, dest_split in split_map.items():
        source_images_dir = os.path.join(source_root, source_split, 'images')
        source_labels_dir = os.path.join(source_root, source_split, 'labels')

        dest_images_dir = os.path.join(destination_root, dest_split, 'images')
        dest_labels_dir = os.path.join(destination_root, dest_split, 'labels')

        # Always create destination directories to ensure the structure exists
        os.makedirs(dest_images_dir, exist_ok=True)
        os.makedirs(dest_labels_dir, exist_ok=True)

        # Copy images
        if os.path.exists(source_images_dir):
            for item in os.listdir(source_images_dir):
                s = os.path.join(source_images_dir, item)
                d = os.path.join(dest_images_dir, item)
                if os.path.isfile(s):
                    shutil.copy2(s, d)
            print(f"Copied images from {source_images_dir} to {dest_images_dir}")
        else:
            print(f"Warning: Source images directory not found: {source_images_dir}. Skipping image copy for this split.")

        # Copy labels
        if os.path.exists(source_labels_dir):
            for item in os.listdir(source_labels_dir):
                s = os.path.join(source_labels_dir, item)
                d = os.path.join(dest_labels_dir, item)
                if os.path.isfile(s):
                    shutil.copy2(s, d)
            print(f"Copied labels from {source_labels_dir} to {dest_labels_dir}")
        else:
            print(f"Warning: Source labels directory not found: {source_labels_dir}. Skipping label copy for this split.")

print("Copying data from Dataset 1...")
copy_yolov8_data(dataset1_path, combined_dataset_path)

print("\nCopying data from Dataset 2...")
copy_yolov8_data(dataset2_path, combined_dataset_path)

print("\nDataset combination complete. Now verifying contents.")
```

```python
import os
import glob

# Re-run the count_unique_classes function to get accurate counts after copying
# Assuming combined_dataset_path is defined from previous cells

def count_unique_classes_after_copy(dataset_path):
    unique_class_ids = set()
    label_dirs = [
        os.path.join(dataset_path, 'train', 'labels'),
        os.path.join(dataset_path, 'val', 'labels')
    ]

    for label_dir in label_dirs:
        if not os.path.exists(label_dir):
            print(f"Warning: Label directory not found: {label_dir}. Skipping.")
            continue

        label_files = glob.glob(os.path.join(label_dir, '*.txt'))
        print(f"Scanning {len(label_files)} label files in {label_dir}...")

        for label_file in label_files:
            with open(label_file, 'r') as f:
                for line in f:
                    try:
                        class_id = int(line.split(' ')[0])
                        unique_class_ids.add(class_id)
                    except (ValueError, IndexError):
                        print(f"Warning: Could not parse class ID from line in {label_file}: {line.strip()}")
                        continue

    sorted_class_ids = sorted(list(unique_class_ids))
    num_unique_classes = len(sorted_class_ids)

    print(f"\n--- Class Analysis Complete (After Copy) ---")
    print(f"Total unique classes found (nc): {num_unique_classes}")
    print(f"Unique class IDs: {sorted_class_ids}")
    if num_unique_classes == 0:
        print("No classes found. Please ensure your label files are correctly formatted and present.")
    else:
        print("\nNow, we will update the `data.yaml` file with these details.")
    return num_unique_classes, sorted_class_ids

# Get class information
num_classes, class_ids = count_unique_classes_after_copy(combined_dataset_path)

# Update data.yaml dynamically
import yaml

# Use the combined_dataset_path variable for consistency
yaml_path = os.path.join(combined_dataset_path, 'data.yaml')

# Load existing data.yaml or create if not exists (should exist from dd6a39fb)
if os.path.exists(yaml_path):
    with open(yaml_path, 'r') as f:
        data = yaml.safe_load(f)
else:
    data = {'path': combined_dataset_path, 'train': 'train/images', 'val': 'val/images', 'nc': 0, 'names': {}}

# Update 'nc' and 'names'
data['nc'] = num_classes
data['names'] = {i: f'class_{i}' for i in range(num_classes)} # Using generic names for now, user can change them

# Save the updated data.yaml
with open(yaml_path, 'w') as f:
    yaml.dump(data, f, sort_keys=False)

print("\n✅ data.yaml imesasishwa kikamilifu na maelezo ya darasa baada ya kunakili data!")
print("------------------------------------")
print(f"nc: {data['nc']}")
print("names:")
for k, v in data['names'].items():
    print(f"  {k}: '{v}'")

# Verify content of the combined dataset after copying
print(f"\n--- Contents of '{combined_dataset_path}' after re-copying and updating data.yaml ---")
!ls -F {combined_dataset_path}
```

```python
import cv2
import matplotlib.pyplot as plt
import os
import random

# Chukua class names zilizosasishwa kutoka kwenye `data` dictionary
# (ambayo imepakiwa kutoka data.yaml kwenye hatua zilizopita)
# Hakikisha 'data' variable bado ipo kwenye kernel state

if 'data' in locals() and 'names' in data:
    class_names_map = data['names']
    print("Class names zilizopakiwa kutoka data.yaml:", class_names_map)
else:
    print("Haikuweza kupata 'data' dictionary au 'names' key. Tafadhali hakikisha data.yaml imepakiwa kwa usahihi.")
    # Fallback to generic names if data.yaml data is not available
    class_names_map = {i: f'class_{i}' for i in range(num_classes)} if 'num_classes' in locals() else {0: 'unknown'}

# Tengeneza rangi za kipekee kwa kila darasa
# Unaweza kurekebisha hii kutoa rangi zaidi au kutumia paleti maalum
bbox_colors = [(random.randint(0, 255), random.randint(0, 255), random.randint(0, 255)) for _ in range(len(class_names_map))]

# Njia za folda za picha na label kwenye validation set
val_images_path = os.path.join(combined_dataset_path, 'val', 'images')
val_labels_path = os.path.join(combined_dataset_path, 'val', 'labels')

# Hakikisha folda zipo
if not os.path.exists(val_images_path):
    print(f"Error: Validation images path not found: {val_images_path}")
    exit()

if not os.path.exists(val_labels_path):
    print(f"Error: Validation labels path not found: {val_labels_path}")
    exit()

# Chagua picha chache za mfano
image_files = [f for f in os.listdir(val_images_path) if f.endswith(('.jpg', '.jpeg', '.png'))]
num_images_to_display = min(3, len(image_files)) # Onyesha max 3 picha

sampled_images = random.sample(image_files, num_images_to_display)

print(f"\nKuonyesha picha {num_images_to_display} za mfano kutoka '{val_images_path}':\n")

for img_file in sampled_images:
    img_path = os.path.join(val_images_path, img_file)
    label_file = os.path.join(val_labels_path, os.path.splitext(img_file)[0] + '.txt')

    img = cv2.imread(img_path) # Soma picha
    if img is None:
        print(f"Warning: Could not read image: {img_path}. Skipping.")
        continue

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) # Badili BGR kwenda RGB kwa matplotlib
    h, w, _ = img.shape

    # Chora bounding boxes
    if os.path.exists(label_file):
        with open(label_file, 'r') as f:
            labels = f.readlines()

        img_with_boxes = img_rgb.copy()

        for label_line in labels:
            try:
                parts = list(map(float, label_line.strip().split()))
                class_id = int(parts[0])
                x_center, y_center, bbox_width, bbox_height = parts[1:]

                # Badili YOLO format kwenda pixel coordinates
                x1 = int((x_center - bbox_width / 2) * w)
                y1 = int((y_center - bbox_height / 2) * h)
                x2 = int((x_center + bbox_width / 2) * w)
                y2 = int((y_center + bbox_height / 2) * h)

                # Pata jina la darasa na rangi
                class_name = class_names_map.get(class_id, f'Class {class_id} (Unknown)')
                color = bbox_colors[class_id % len(bbox_colors)] # Chagua rangi kutoka kwenye orodha

                # Chora mstatili
                cv2.rectangle(img_with_boxes, (x1, y1), (x2, y2), color, 2)

                # Weka jina la darasa
                cv2.putText(img_with_boxes, class_name, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.7, color, 2)

            except Exception as e:
                print(f"Warning: Error parsing label line '{label_line.strip()}' in '{label_file}': {e}")
                continue
    else:
        img_with_boxes = img_rgb.copy()
        cv2.putText(img_with_boxes, "No Labels Found", (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)

    # Onyesha picha
    plt.figure(figsize=(10, 10))
    plt.imshow(img_with_boxes)
    plt.title(f"Image: {img_file}")
    plt.axis('off')
    plt.show()
```
